```python
# ----------------------------
# Load data
# First column = Treatment
# Remaining columns = features
# ----------------------------

data = pd.read_csv("Leukemia_GSE9476.csv")

y = data.iloc[:,1]      # treatment labels
X = data.iloc[:,2:]     # features

# ----------------------------
# Scale features
# PCA requires scaling
# ----------------------------

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ----------------------------
# PCA
# ----------------------------

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1","PC2"]
)

pca_df["Treatment"] = y.values

# ----------------------------
# Plot PCA by group
# ----------------------------

for group in pca_df["Treatment"].unique():

    subset = pca_df[
        pca_df["Treatment"] == group
    ]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=group
    )

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)"
)

plt.legend()
plt.show()

# ----------------------------
# Classification model
# ----------------------------

rf = RandomForestClassifier(
    n_estimators=1000,
    random_state=42
)

rf.fit(X_scaled, y)

perm = permutation_importance(
    rf,
    X_scaled,
    y,
    n_repeats=30,
    random_state=42,
    scoring="accuracy"
)
# Overall feature importance
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": perm.importances_mean,
    "Std": perm.importances_std
})

print("\nOverall top features:")
print(importance.head(20))

# ----------------------------
# Feature importance by group
# One-vs-rest
# ----------------------------

group_results = {}

for group in y.unique():

    y_binary = (y == group).astype(int)

    rf_group = RandomForestClassifier(
        n_estimators=1000,
        random_state=42
    )

    rf_group.fit(
        X_scaled,
        y_binary
    )

    imp = pd.DataFrame({
        "Feature": X.columns,
        "Importance":
        rf_group.feature_importances_
    })

    imp = imp.sort_values(
        "Importance",
        ascending=False
    )

    group_results[group] = imp

print("\nOverall top features")

print(importance.head(20))
```
