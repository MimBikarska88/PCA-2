```python
import pandas as pd
import pingouin as pg
import numpy as np
from skbio.stats.distance import permanova, DistanceMatrix
from sklearn.metrics import pairwise_distances
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier, NeighborhoodComponentsAnalysis
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance
from sklearn.datasets import make_classification
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score, f1_score
import plotly.express as px
from statsmodels.stats.outliers_influence import variance_inflation_factor

n_neighbors = 3
random_state = 0
```

### Mice Proteins Expressions ###
The data set consists of the expression levels of 77 proteins/protein modifications that produced detectable signals in the nuclear fraction of cortex. There are 38 control mice and 34 trisomic mice (Down syndrome), for a total of 72 mice. In the experiments, 15 measurements were registered of each protein per sample/mouse. Therefore, for control mice, there are 38x15, or 570 measurements, and for trisomic mice, there are 34x15, or 510 measurements. The dataset contains a total of 1080 measurements per protein. Each measurement can be considered as an independent sample/mouse.

The eight classes of mice are described based on features such as genotype, behavior and treatment. According to genotype, mice can be control or trisomic. According to behavior, some mice have been stimulated to learn (context-shock) and others have not (shock-context) and in order to assess the effect of the drug memantine in recovering the ability to learn in trisomic mice, some mice have been injected with the drug and others have not.

Attribute Information
- Mouse ID

- Values of expression levels of 77 proteins; the names of proteins are followed by N indicating that they were measured in the nuclear fraction. For example: DYRK1A_n

- Genotype: control (c) or trisomy (t)

- Treatment type: memantine (m) or saline (s)

- Behavior: context-shock (CS) or shock-context (SC)

- Class: c-CS-s, c-CS-m, c-SC-s, c-SC-m, t-CS-s, t-CS-m, t-SC-s, t-SC-m

 __Task: The aim is to identify subsets of proteins that are discriminant between the classes.__

### Approach 1 ###
I want to see if proteins differ significantly accross different classes 
$H_0$ - There is no significant difference in protein expression in different classes of mice and
$H_1$ - There is significant difference in protein expression in different classes of mice

If $H_0$ is rejected, I want to know which proteins are determinant in different groups

__Type of data:__<br>
   Response variable: protein expressions ( multivariate as there is a large number of columns for each)<br>
   Predictor variable: class -> nominal<br>
__Design study:__ <br>
 Non-Parametric test is more appropriate as data is not normally distributed and classes are independent<br>
__Number of groups:__ <br> more than 2 (8 in total)

```python
mice_protein_data = mice_protein_data.drop(columns = ['Genotype','Treatment','Behavior'])
```

```python
mice_protein_data = mice_protein_data.dropna()
X = mice_protein_data.iloc[:, 1:78]
y = mice_protein_data['class'].reset_index(drop=True)
ids = y.index.tolist()
dist_array = pairwise_distances(
    X,
    metric='braycurtis'
)
dm = DistanceMatrix(dist_array, ids=ids)
print(X.shape)
print(y.shape)
results = permanova(dm, y, permutations=999)
print(results)
```

### Find out prominent proteins globally using Kruskal test  ###
After we prove that there is significant difference between the protein expressions of different classes we want to find the most significant proteins globally for all classes. This is only the first step of this analysis, afterwards, we will set a threshold for those with smalles p-values and see how definitive are they per class.



```python
alpha = 0.05
p_values = []

#do it for all proteins 
for protein in X:
    #kruskal-wallis works with groups, but won't group internally
    groups = []
    for class_label in y.unique():
        values = mice_protein_data.loc[mice_protein_data['class']==class_label, [protein]]
        groups.append(values)
    statistic, pvalue = stats.kruskal(*groups)
    print(f"{protein}: Statistic {statistic} P-value {pvalue}\n")
    
```

It seems that our Kruskal test did not provide results that are good for our goal, so we can try a few other approaches

### Check For Multicollinearity ###
We see that just a few values have positive coefficients, so we despite our model's performance we can check for collinear values. What does collinearity mean? 
It occurs when two or more features in a regression model are correlated <br><br>
__Inflated Standard Errors__ : Coefficients become less statistically significant <br><br>
__Unstable Coefficients__ : Small changes in data can lead to large coefficients variations <br><br>
__Reduced Model Interprettability__: Difficult interpeting the impact of individual values<br>
<br>
How to detect it?<br><br>
__Correlation Matrix__: Look for high correlations between predictors ( > 0.9 )<br><br>
__Variance Inflation Factors__:  A VIF > 10 suggests multicollinearity<br><br>
__Condition Index__: Values > 30 suggest multicollinearity

```python
#CI
normalized_features = (X - X.mean()) / X.std() 
cov_matrix = np.cov(normalized_features.T)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix) 
condition_indices = np.sqrt(eigenvalues.max()/eigenvalues)
ci_data = pd.DataFrame({
    'Condition Index':condition_indices
})
ci_data
```

```python
#VIF measures how much the variance of a regression coefficient is inflated due to multicolinearity 
#how much of its information is contained in other variables
vif_data = pd.DataFrame()
vif_data['Feature'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
pd.set_option('display.max_rows', None)
vif_data
```

```python
#Correlation matrix
X_corr = X.corr().abs()
fig = px.imshow(X_corr,  width=1000, height=1000)
fig.show()

upper = X_corr.where(np.triu(np.ones(X_corr.shape), k=1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > 0.75)]
X.drop(to_drop, axis=1, inplace=True)

#Solution: Leave just 1 out of each 2 correlated 

X_corr = X.corr().abs()
fig = px.imshow(X_corr,  width=1000, height=1000)
fig.show()

```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, stratify=y, random_state=random_state
)
```

```python

clf = RandomForestClassifier(random_state = random_state)
clf.fit(X_train, y_train)
result = permutation_importance(clf, X_test, y_test, n_repeats = 10, random_state = random_state)

#define a threshold
#check whether mean > std

for i in range(len(result.importances_mean)):
    gene = X.columns[i]
    print(f"Feature {gene}: {result.importances_mean[i]}")
    
y_pred = clf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Macro F1:", f1_score(y_test, y_pred, average="macro"))
```

```python
feature_importances_df = pd.DataFrame(
    clf.feature_importances_,
    index = X.columns
)
print(feature_importances_df.head(15))
```

We see that in the upper right corner, in the middle and at the end we have squares decreasing in width which indicate collinearity. Let's check the other methods  before we proceed.

```python
# Reduce dimension to 2 with PCA
#n_components and n_neighbours is number of classes - 1 = 7
pca = make_pipeline(StandardScaler(), PCA(n_components=7, random_state=random_state))

# Reduce dimension to 2 with LinearDiscriminantAnalysis
lda = make_pipeline(StandardScaler(), LinearDiscriminantAnalysis(n_components=2))

# Reduce dimension to 2 with NeighborhoodComponentAnalysis
nca = make_pipeline(
    StandardScaler(),
    NeighborhoodComponentsAnalysis(n_components=7, random_state=random_state),
)

# Use a nearest neighbor classifier to evaluate the methods
knn = KNeighborsClassifier(n_neighbors=7)


# Make a list of the methods to be compared
dim_reduction_methods = [("PCA", pca), ("LDA", lda), ("NCA", nca)]

le = LabelEncoder()
y_encoded = le.fit_transform(y)

for i, (name, model) in enumerate(dim_reduction_methods):
    plt.figure()
    # plt.subplot(1, 3, i + 1, aspect=1)

    # Fit the method's model
    model.fit(X_train, y_train)

    # Fit a nearest neighbor classifier on the embedded training set
    knn.fit(model.transform(X_train), y_train)

    # Compute the nearest neighbor accuracy on the embedded test set
    acc_knn = knn.score(model.transform(X_test), y_test)

    # Embed the data set in 2 dimensions using the fitted model
    X_embedded = model.transform(X)

    # Plot the projected points and show the evaluation score
    plt.scatter(X_embedded[:, 0], X_embedded[:, 1], c=y_encoded, s=30, cmap="Set1")
    plt.title(
        "{}, KNN (k={})\nTest accuracy = {:.2f}".format(name, 7, acc_knn)
    )
plt.show()
```
